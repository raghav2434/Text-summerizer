## Step 1: Install & Load Required Libraries

```python
# Install dependencies (uncomment if running in Google Colab)
# !pip install -q transformers torch nltk datasets streamlit

import nltk
import torch
import transformers
from text_summarizer import TextCleaner, ExtractiveSummarizer, AbstractiveSummarizer, SummaryEvaluator, DatasetLoader

print(f"NLTK Version: {nltk.__version__}")
print(f"PyTorch Version: {torch.__version__}")
print(f"Transformers Version: {transformers.__version__}")
```

## Step 2: Define Sample Input Article

```python
sample_text = """
Scientists at the Global Energy Institute have announced a major breakthrough in solar energy storage technology. 
The team has developed a novel hybrid battery system that utilizes abundant, non-toxic organic materials capable of storing solar power for up to six months with minimal efficiency loss. 
Traditional solar energy systems often struggle with energy storage during night hours or prolonged cloudy periods. 
Existing lithium-ion batteries are costly to produce and present environmental challenges during disposal. 
The new organic flow battery overcomes these limitations by using liquid electrolytes derived from plant polymers. 
In testing, the prototype sustained over 10,000 charge cycles while maintaining 98% capacity. 
Industry experts believe this technology could accelerate the global transition to clean energy, making off-grid solar power reliable and affordable for remote communities worldwide. 
Commercial production is slated to begin within the next two years following final safety certifications.
"""

cleaner = TextCleaner()
sentences = cleaner.split_into_sentences(sample_text)
print(f"Total Sentences in Sample: {len(sentences)}")
print(f"Total Words in Sample: {len(cleaner.tokenize_words(sample_text))}")
```

## Step 3: Extractive Summarization (NLP / Sentence Scoring)

```python
ranker = ExtractiveSummarizer(cleaner=cleaner)
extractive_result = ranker.summarize(sample_text, num_sentences=3)

print("=== EXTRACTIVE SUMMARY (NLTK) ===")
print(extractive_result["summary"])
print("\nSelected Sentences:")
for s in extractive_result["sentences"]:
    print(f" - {s}")
```

## Step 4: Abstractive Summarization (Hugging Face AI)

```python
abstractive_model = AbstractiveSummarizer()
abstractive_result = abstractive_model.summarize(sample_text, max_length=90, min_length=30)

print("=== ABSTRACTIVE AI SUMMARY (Transformers) ===")
print(abstractive_result["summary"])
print(f"Model Used: {abstractive_result['model_used']}")
print(f"Method: {abstractive_result['method']}")
```

## Step 5: Benchmark Datasets (CNN/DailyMail & XSum)

```python
dl = DatasetLoader()
articles = dl.get_all_articles()
print(f"Loaded {len(articles)} benchmark articles from dataset.")

for art in articles:
    print(f"\nID: {art['id']} | Category: {art['category']} | Title: {art['title']}")
    eval_res = SummaryEvaluator(cleaner=cleaner).evaluate(art['text'], art['reference_summary'])
    print(f"Original Words: {eval_res['original_word_count']} -> Reference Summary Words: {eval_res['summary_word_count']} (Compression: {eval_res['compression_ratio_percent']}%")
```

## Step 6: Comparative Metrics Evaluation

```python
evaluator = SummaryEvaluator(cleaner=cleaner)
ext_eval = evaluator.evaluate(sample_text, extractive_result["summary"])
abs_eval = evaluator.evaluate(sample_text, abstractive_result["summary"])

import pandas as pd
comparison_df = pd.DataFrame([
    {
        "Method": "Extractive (NLTK)",
        "Summary Word Count": ext_eval["summary_word_count"],
        "Compression %": ext_eval["compression_ratio_percent"],
        "Reading Time Saved (s)": ext_eval["reading_time_saved_sec"]
    },
    {
        "Method": "Abstractive (Hugging Face AI)",
        "Summary Word Count": abs_eval["summary_word_count"],
        "Compression %": abs_eval["compression_ratio_percent"],
        "Reading Time Saved (s)": abs_eval["reading_time_saved_sec"]
    }
])

display(comparison_df)
```
